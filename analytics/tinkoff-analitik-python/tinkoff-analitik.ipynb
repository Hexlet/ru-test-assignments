```
#проверим еще раз, не должно быть дубликатов по составному ключу 'user_id-utilization_date'
# (один и тот же клиент не может регистрироваться несколько раз, да ещё в одно и то же время)
df_initial['id-date'] = df_initial['user_id'].map(str) + df_initial['utilization_date'].map(str) # создаем
# составной уникальный ключ
df_initial.head(3)
```

```
pd.DataFrame([{'id-date': len(df_initial['id-date'].value_counts())
              }], columns = ['id-date'], index = ['Количество'])
```

```
df_initial.drop_duplicates(subset=['id-date'], inplace = True) # удалим дубликаты, их действительно очень много
print('Таблица данных: строк, столбцов', df_initial.shape)
```

```
df_initial['month_year'] = pd.to_datetime( df_initial[ 'utilization_date' ]).dt.to_period( 'M' )
# добавим столбец месяц год, чтобы можно было сделать свод по месяцам
```

```
df_initial.head(3)
```

```
from matplotlib import pyplot as plt # подключим графическую библиотеку pyplot  и назовем ее коротко plt

df_initial.groupby(['month_year'])['user_id'].nunique().plot.bar() # подсчитаем количество клиентов по месяцам
#здесь используется достаточно специфическая конcтрукция, мы действуем на df_initial несколькими методами, перечисляя их через "."

plt.xlabel('по месяцам') # подпишем оси
plt.ylabel('количество клиентов')
# можем и сложнее графики строить, но это отдельная задача, выйдем за рамки занятия
```

```
# как меняется стоимость привлечения?
df_initial.groupby(['month_year'])['cost_of_utilization'].sum().plot.bar()
# подсчитаем стоимость привлечения по месяцам

plt.xlabel('по месяцам') # подпишем оси
plt.ylabel('стоимость привлечения')
# можем и сложнее графики строить, но это отдельная задача, выйдем за рамки занятия
```

Мы видим, что вроде особых изменений нет, ну меняются затраты на привлечение, но и клиентов тоже вместе с этим прибавляется.

```
#проверим меняется ли прогноз LTV
df_initial.groupby(['month_year'])['LTV_forecast'].sum().plot.bar() # подсчитаем прогноз LTV по месяцам

plt.xlabel('по месяцам') # подпишем оси
plt.ylabel('Прогноз LTV')
# можем и сложнее графики строить, но это отдельная задача, выйдем за рамки занятия
```

Надо что-то придумать, просто сказать, что "всё по-прежнему" как-то странно, зачем аналитик тогда?

```
monthly = df_initial.groupby(['month_year']).agg({'user_id': ['count'], 'cost_of_utilization': ['sum'],
                                       'LTV_forecast': ['sum']}) # сочиним новую, сводную табличку
monthly.columns = monthly.columns.map(' '.join).str.strip() # упростим мультииндексы (сложные двухуровневые заголовки)
monthly = monthly.reset_index(level='month_year')  # переведем индекс в обычный столбец
monthly ['month_year'] = monthly ['month_year'].map(str)  # изменим формат, чтобы графики можно было строить
print (monthly)
```

Введем пару новых метрик, может они прояснят ситуацию. Больше метрик Богу метрик!

```
monthly ['средняя стоимость привлечения'] = monthly['cost_of_utilization sum'] /  monthly['user_id count'] # новая метрика по привлечению
monthly ['средний прогноз LTV'] = monthly['LTV_forecast sum'] /  monthly['user_id count'] # новая метрика по прогнозу LTV
monthly.drop(['user_id count',  'cost_of_utilization sum',  'LTV_forecast sum'], axis = 1, inplace = True) # старые выкинем
print (monthly)
```

Кажется что-то прояснилось. Построим графики, мы же аналитики!

```
plt.plot( 'month_year', 'средняя стоимость привлечения', data = monthly, color = 'skyblue', linewidth = 4)
plt.plot( 'month_year', 'средний прогноз LTV', data = monthly, color = 'olive', linewidth = 4)

plt.legend()
```
